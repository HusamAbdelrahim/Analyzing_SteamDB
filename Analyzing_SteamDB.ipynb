```python

df = pd.read_csv("SteamDB_TopFiveGames.csv")
df_tidy1 = pd.melt(
    df,
    id_vars = ["app_id", "game_name", "developer","release_year", "Positive_reviews", "Negative_reviews", "24-hour peak", "all time peak", "gain_may", "gain_june", "gain_july", "gain_30_days"],
    value_vars= ["gain1%", "gain2%", "gain3%", "gain4%"],
    var_name= "Gain Type",
    value_name= "Percentage"
)

total_sum = 0
count = 0

for index, row in df.iterrows():
    total_sum += row ['24-hour peak']
    count += 1

average = total_sum / count if count > 0 else 0

mask = df["game_name"] == "Counter-Strike 2"

cs2_peak = df.loc[mask, "24-hour peak"].iloc[0]
cs2_alltime = df.loc[mask, "all time peak"].iloc[0]

cs2_pct = round(cs2_peak/cs2_alltime * 100, 1)
print(cs2_pct)

df_calculation = df["Peak %"] = (df["24-hour peak"] / df["all time peak"] * 100).round(1)

# cs24 = 1277861
# cspeak = 1862531
# percentage2  = (cs24 / cspeak) * 100
# print(percentage2)

#positive_reviews_x = df("Positive reivews")
#negative_reviews_y = df("Negative reviews")

# First Graph

plt.figure(figsize=(12,6))

plt.bar(df["game_name"], df["Positive_reviews"], label = "Positive", color = "skyblue")
plt.bar(df["game_name"], df["Negative_reviews"], bottom=df["Positive_reviews"], label = "Negative", color = "salmon")

plt.ylabel("Positive reviews (%)")
plt.title("Positive review share of the game")
plt.xticks(rotation = 45, ha = "right")
plt.tight_layout()
plt.legend()
plt.show()

# 2nd Graph

plt.figure(figsize=(12,6))
plt.plot(df["gain_may"], label = "May", color = "red")
plt.plot(df["gain_june"], label = "June", color = "blue")
plt.plot(df["gain_july"], label = "July", color = "purple")
plt.plot(df["gain_30_days"], label = "30 Days", color = "pink" )
plt.title("Gain Progress")
plt.ylabel("gain")
plt.legend()
plt.show()

# 3rd Graph

wide = df.set_index("game_name")[["24-hour peak", "all time peak"]]
ax = wide.plot.bar(figsize = (12,6), logy = True)
ax.set_ylabel("Players (log scale)")
plt.tight_layout()
plt.show()

print(df_calculation)
print(f"Average: {average}")
df_tidy1.head(5)
```
